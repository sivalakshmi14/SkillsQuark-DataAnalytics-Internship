```python

import pandas as pd

data = {
    "Order_ID": range(1, 16),
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop", "Phone",
        "Tablet", "Laptop", "Phone", "Tablet", "Laptop",
        "Phone", "Tablet", "Laptop", "Phone", "Tablet"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics"
    ],
    "Region": [
        "North", "South", "East", "West", "South",
        "East", "North", "West", "South", "North",
        "East", "West", "South", "North", "East"
    ],
    "Sales": [
        50000, 25000, 18000, 55000, 27000,
        20000, 52000, 26000, 19000, 150000,
        28000, 22000, 48000, 24000, 21000
    ],
    "Profit": [
        8000, 4000, 2500, 9000, 4500,
        3000, 8500, 4200, 2800, 30000,
        4700, 3200, 7500, 3900, 3100
    ],
    "Quantity": [
        5, 10, 7, 6, 12, 8, 5, 11, 9, 4,
        10, 8, 5, 9, 7
    ]
}

df = pd.DataFrame(data)

display(df.head())
print("Total records:", len(df))
print("Total sales:", df["Sales"].sum())
print("Dataset created successfully!")
```

```python

print(df["Sales"].tolist())
print("Total Sales:", df["Sales"].sum())
```

```python

import plotly.express as px

product_sales = df.groupby("Product", as_index=False)["Sales"].sum()

fig = px.bar(
    product_sales,
    x="Product",
    y="Sales",
    title="Total Sales by Product",
    color="Product",
    text="Sales"
)

fig.update_layout(
    xaxis_title="Product",
    yaxis_title="Total Sales"
)

fig.show()
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output

region_filter = widgets.Dropdown(
    options=["All Regions"] + sorted(df["Region"].unique().tolist()),
    description="Region:"
)

output = widgets.Output()

def update_dashboard(change=None):
    with output:
        clear_output(wait=True)

        if region_filter.value == "All Regions":
            filtered_df = df.copy()
        else:
            filtered_df = df[df["Region"] == region_filter.value]

        print("Selected Region:", region_filter.value)
        print("Total Sales:", filtered_df["Sales"].sum())
        print("Total Profit:", filtered_df["Profit"].sum())

        fig = px.bar(
            filtered_df.groupby("Product", as_index=False)["Sales"].sum(),
            x="Product",
            y="Sales",
            title="Sales by Product and Region",
            color="Product"
        )
        fig.show()

region_filter.observe(update_dashboard, names="value")
display(region_filter, output)
update_dashboard()
```

```python

from IPython.display import display
import plotly.express as px

fig = px.bar(
    df.groupby("Product", as_index=False)["Sales"].sum(),
    x="Product",
    y="Sales",
    title="Total Sales by Product",
    color="Product"
)

display(fig)
```

```python

import plotly.graph_objects as go
from IPython.display import display

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = len(df)

fig = go.Figure()

fig.add_trace(go.Indicator(
    mode="number",
    value=total_sales,
    title={"text": "Total Sales"}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=total_profit,
    title={"text": "Total Profit"},
    domain={"x": [0.5, 1], "y": [0, 1]}
))

fig.update_layout(
    title="Sales Performance Dashboard",
    grid={"rows": 1, "columns": 2, "pattern": "independent"}
)

fig.show()

print("Total Orders:", total_orders)
```

```python

import plotly.express as px

profit_data = df.groupby(
    "Product", as_index=False
)["Profit"].sum()

fig = px.pie(
    profit_data,
    names="Product",
    values="Profit",
    title="Profit Distribution by Product",
    hole=0.4
)

fig.show()
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output

product_filter = widgets.Dropdown(
    options=["All Products"] + sorted(df["Product"].unique().tolist()),
    description="Product:"
)

region_filter_table = widgets.Dropdown(
    options=["All Regions"] + sorted(df["Region"].unique().tolist()),
    description="Region:"
)

table_output = widgets.Output()

def show_sales_table(change=None):
    with table_output:
        clear_output(wait=True)

        filtered = df.copy()

        if product_filter.value != "All Products":
            filtered = filtered[
                filtered["Product"] == product_filter.value
            ]

        if region_filter_table.value != "All Regions":
            filtered = filtered[
                filtered["Region"] == region_filter_table.value
            ]

        display(filtered)
        print("Filtered records:", len(filtered))

product_filter.observe(show_sales_table, names="value")
region_filter_table.observe(show_sales_table, names="value")

display(product_filter, region_filter_table, table_output)
show_sales_table()
```

```python

print("INTERACTIVE BI DASHBOARD - KEY FINDINGS")
print("----------------------------------------")

print("Total Orders:", len(df))
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Quantity Sold:", df["Quantity"].sum())

print("\nSales by Product:")
display(
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Region:")
display(
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nProfit by Product:")
display(
    df.groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
```
