```python

import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50000, 25000, 18000],
    "Profit": [8000, 4000, 2500]
}

df = pd.DataFrame(data)
display(df)
```

```python

print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Best-Selling Product:",
      df.loc[df["Sales"].idxmax(), "Product"])
```

```python

import matplotlib.pyplot as plt

plt.bar(df["Product"], df["Sales"])
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python

best_product = df.loc[df["Sales"].idxmax(), "Product"]
low_product = df.loc[df["Sales"].idxmin(), "Product"]

print("BUSINESS INSIGHTS")
print("1. Highest sales product:", best_product)
print("2. Lowest sales product:", low_product)

print("\nRECOMMENDATIONS")
print("1. Focus on products with higher sales.")
print("2. Study why some products have lower sales.")
print("3. Monitor sales and profit regularly.")
```

```python

report = f"""
BUSINESS INSIGHTS REPORT

1. Total Sales: {df["Sales"].sum()}
2. Total Profit: {df["Profit"].sum()}
3. Best-Selling Product: {df.loc[df["Sales"].idxmax(), "Product"]}
4. Lowest-Selling Product: {df.loc[df["Sales"].idxmin(), "Product"]}

Business Recommendations:
- Focus on high-performing products.
- Improve marketing for low-selling products.
- Monitor sales and profit regularly.

Tools Used: Python, Pandas, Matplotlib, Google Colab
"""

print(report)

with open("Task5_Business_Insights_Report.txt", "w") as file:
    file.write(report)
```
