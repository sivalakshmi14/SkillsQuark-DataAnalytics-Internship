```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone",
                "Tablet", "Laptop", "Phone", "Tablet", "Laptop"],
    "Sales": [50000, 25000, 18000, 55000, 27000,
              20000, 52000, 26000, 19000, 150000],
    "Profit": [8000, 4000, 2500, 9000, 4500,
               3000, 8500, 4200, 2800, 30000],
    "Quantity": [5, 10, 7, 6, 12, 8, 5, 11, 9, 4],
    "Region": ["North", "South", "East", "West", "South",
               "East", "North", "West", "South", "North"]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
display(df)
```

```python

print("Dataset Shape (Rows, Columns):")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nStatistical Summary:")
display(df.describe())
```

```python

print("Missing Values in Each Column:")
display(df.isnull().sum())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())
```

```python

product_sales = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(8, 5))
product_sales.plot(kind="bar")
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

numeric_data = df[["Sales", "Profit", "Quantity"]]

correlation_matrix = numeric_data.corr()

print("Correlation Matrix:")
display(correlation_matrix)

plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Sales, Profit and Quantity")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Product", y="Sales")

plt.title("Sales Outlier Detection by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```
